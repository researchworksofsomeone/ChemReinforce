```python
import subprocess, threading
def _run_pinggy():
    import os
    os.system("pkill -f 'ssh -p 443'")
    proc = subprocess.Popen(
        ["ssh", "-p", "443", "-o", "StrictHostKeyChecking=no", "-o", "ServerAliveInterval=30", "-R0:localhost:22", "<PINGGY_TOKEN>+tcp@free.pinggy.io"],
        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    for line in proc.stdout: print(line, end="", flush=True)
t = threading.Thread(target=_run_pinggy, daemon=True)
t.start()
print("Pinggy SSH starting...")
```
